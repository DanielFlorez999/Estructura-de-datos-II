```python
#Diccionarios
d1 = {
  "Nombre": "Daniel",
  "Edad": 22,
  "Documento": 1000835256
}
print(d1)
```

```python
#Otra forma de hacer un dccionario
d2 = dict([
      ('Nombre', 'Daniel'),
      ('Edad', 22),
      ('Documento', 1000835256),
])
print(d2)
```

```python
#Forma final de hacer un diccionario
d3 = dict(Nombre='Daniel',
          Edad=22,
          Documento=1000835256)
print(d3)
```

```python
#Funcion de acceder al diccionario con get o con corchetes
d1 = {
  "Nombre": "Daniel",
  "Edad": 22,
  "Documento": 1000835256
}

print(d1['Nombre'])
print(d1.get('Nombre'))

print(d1['Edad'])
print(d1.get('Edad'))
```

```python
#Funcion de modificar el diccionario con corchetes
d1 = {
  "Nombre": "Daniel",
  "Edad": 22,
  "Documento": 1000835256
}

d1['Nombre'] = "Laura"
print(d1)
```

```python
#Funcion de modificar el diccionario con corchetes y si el key o elemento no existe el diccionario lo añade automaticamente
d1 = {
  "Nombre": "Daniel",
  "Edad": 22,
  "Documento": 1000835256
}
d1['Direccion'] = "Calle 123"
print(d1)
```

```python
#Iterar un diccionario solo imprime las key

d1 = {
  "Nombre": "Daniel",
  "Edad": 22,
  "Documento": 1000835256
}

for x in d1:
  print(x)
```

```python
#Iterar un diccionario solo imprime los value

d1 = {
  "Nombre": "Daniel",
  "Edad": 22,
  "Documento": 1000835256
}

for x in d1:
  print(d1[x])
```

```python
#Iterar un diccionario e imprimir la key y el value po coordenadas

d1 = {
  "Nombre": "Daniel",
  "Edad": 22,
  "Documento": 1000835256
}

for x, y in d1.items():
    print(x, y)
```

```python
#Diccionarios anidados

anidado1 = {"a": 1, "b": 2}
anidado2 = {"a": 1, "b": 2}
d = {
  "anidado1" : anidado1,
  "anidado2" : anidado2
}
print(d)
```

```python
#METODOS DE LOS DICCIONARIOS
#Metodo CLEAR, elimina todo lo que hay dentro del diccionario

d = {'a': 1, 'b': 2}
d.clear()
print(d)
```

```python
#Metodo GET consultar un value de una key determinada, y con un parametro opcional por si el parametro no se encuentra en el diccionario

d = {'a': 1, 'b': 2}
print(d.get('a')) #1
print(d.get('z', 'No encontrado'))
```

```python
#Metodo ITEMS devuelve una lista con los keys y values del diccionario. Si se convierte en list se puede indexar como si de una lista normal se tratase, siendo los primeros elementos las key y los segundos los value.

d = {'a': 1, 'b': 2}
it = d.items()
print(it)             #dict_items([('a', 1), ('b', 2)])
print(list(it))       #[('a', 1), ('b', 2)]
print(list(it)[0][0]) #a
```

```python
#El método KEYS devuelve una lista con todas las keys del diccionario.

d = {'a': 1, 'b': 2}
k = d.keys()
print(k)       #dict_keys(['a', 'b'])
print(list(k)) #['a', 'b']
```

```python
#El método VALUES devuelve una lista con todos los values o valores del diccionario.

d = {'a': 1, 'b': 2}
print(list(d.values())) #[1, 2]
```

```python
#El método pop() busca y elimina la key que se pasa como parámetro y devuelve su valor asociado. Daría un error si se intenta eliminar una key que no existe.
#También se puede pasar un segundo parámetro que es el valor a devolver si la key no se ha encontrado. En este caso si no se encuentra no habría error.

d = {'a': 1, 'b': 2}
d.pop('a')
print(d) #{'b': 2}

e = {'a': 1, 'b': 2}
e.pop('c', -1)
print(e) #{'a': 1, 'b': 2}
```

```python
# El método POPITEM elimina de manera aleatoria un elemento del diccionario.

d = {'a': 1, 'b': 2}
d.popitem()
print(d)
#{'a': 1}
```

```python
#El método update() se llama sobre un diccionario y tiene como entrada otro diccionario. Los value son actualizados y si alguna key del nuevo diccionario no esta, es añadida.

d1 = {'a': 1, 'b': 2}
d2 = {'a': 0, 'd': 400}
d1.update(d2)
print(d1)
#{'a': 0, 'b': 2, 'd': 400}
```
