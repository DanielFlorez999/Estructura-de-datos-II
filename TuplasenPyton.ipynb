```python
#tuplas
t = (1, 1, 2, 3, 3, 3)
print(t)
```

```python
#Tupla con ciclo for

colores = ("Rojo", "Verde", "Azul")

for color in colores:
  print(color)
```

```python
t = ([1, 2][1, 3][1, 4])
print(t[0])
print(t[0][1])
```

```python
#anidacion de tuplas
t1 = (1, 2, 3)
t2 = (4, 5, 6)

resultado = t1 + t2
print(resultado)
```

```python
#Repeticion de tuplas
t = ("python",)*3
print(t)
```

```python
#Pertenencia de tuplas

t = (10, 20, 30, 40)

print(20 in t)
print(50 in t)
```

```python
#Desempaquetado de tuplas

persona = ("Nicolas", 22, "Colombia")

nombre, edad, pais = persona

print(nombre)
print(edad)
print(pais)
```

```python
#Metodos disponibles count y index

t = (1, 2, 3, 4)

print(t.count(2))
print(t.index(3))
```

```python
#Funciones utiles

t = (10, 20, 30, 40)

print(len(t))
print(max(t))
print(min(t))
print(sum(t))
print(sorted(t))

```

```python
#conversion entre lista y tupla

lista = [1, 2, 3, 4]
tupla = tuple(lista)

print(tupla)

nueva_lista = list(tupla)
print(nueva_lista)
```

```python
#Tuplas como registros

persona = ("Nicolas", 22, "Colombia")

print("nombre: ", persona[0])
print("edad: ", persona[1])
print("pais: ", pais[2])
```

```python
coordenadas = {}

punto = (10, 20)
coordenadas[punto] = "Punto 1"

print(coordenadas)
```

```python
#devolviendo multiples valores

def operaciones(a, b):
  return (a + b, a - b, a * b, a / b)

suma, resta, multiplicacion, division = operaciones(10, 5)

print(suma)
print(resta)
print(multiplicacion)
print(division)
```

```python
#Iteracion con enumerate()

frutas = ["manzana", "banana", "cereza"]

for indice, fruta in enumerate(frutas):
  print(indice, fruta)
```

```python
#intercambio de valores

a, b = 10, 20
a, b = b, a
print(a, b)
```
