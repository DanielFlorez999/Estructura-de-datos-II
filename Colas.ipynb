Colas en python

```python
#Colas
cola = []
cola.append("A")
cola.append("B")
cola.append("C")
print(cola)

pop1 = cola.pop(0)
pop2 = cola.pop()
print(pop1)
print(cola)

print("se elimino ",pop1)
print("se elimino ",pop2)
```

```python
#Cola popleft

from collections import deque
cola = deque(['A', 'B', 'C'])
print(cola)
cola.append("D")
print(cola)

primero = cola.popleft()
print("Se elimino ",primero)
print(cola)
```

```python
#Colas dobles
from collections import deque
cola_doble = deque(['A', 'B', 'C'])
cola_doble.append("D")
cola_doble.appendleft("E")
print(cola_doble)

cola_doble.pop()
primero = cola_doble.popleft()

print(cola_doble)
```

```python
#Cola con priorityQueue

from queue import PriorityQueue
cola = PriorityQueue()

cola.put((1, "Paciente gravedad 1"))
cola.put((3, "Paciente medio"))
cola.put((2, "Paciente leve 3"))

print("Atendido", cola.get())
print("Atendido", cola.get())
print("Atendido", cola.get())
```

```python
#Cola circular

from collections import deque

jugadores = deque(["A", "B", "C", "D"])

for i in range(10):
  turno = jugadores.popleft()
  print(turno)
  jugadores.append(turno)

print("Cola final", jugadores)
```

```python
#Cola ejercicio practico

from collections import deque

eventos = deque()

while True:
  print("Bienvenido al sistema de eventos")
  print("1. Agregar evento")
  print("2. Atender evento")
  print("3. Eliminar eventos")
  print("4. Salir")

  opc = int(input("Ingrese una opcion: "))

  if opc == 1:
    evento = input("Ingrese el evento: ")
    eventos.append(evento)
    print("Evento agregado")

  elif opc == 2:
    if len(eventos) == 0:
      print("No hay eventos")
  elif opc == 3:
      print("Evento eliminado")
  else:
      break
```
